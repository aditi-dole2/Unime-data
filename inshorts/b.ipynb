```python
import pandas as pd

# The given list of named entities
data = [
    [('million', 'CARDINAL')],
    [('angeles', 'GPE')],
    [],
    [('carolina', 'GPE'), ('tabasco', 'ORG')],
    [('amy', 'PERSON'), ('cooper', 'ORG'), ('franklin', 'ORG')]
]

# Flatten the list of tuples
flattened_data = [item for sublist in data for item in sublist]

# Create a DataFrame with columns "entity" and "label"
df = pd.DataFrame(flattened_data, columns=['entity', 'label'])

# Display the DataFrame
print(df)
```
