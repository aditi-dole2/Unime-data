```python
import pandas as pd
```

```python
auth = pd.read_excel("authorrs.xlsx")
auth['authors']=auth['authors'].fillna('None')
auth.head()
auth['authors'] =auth['authors'].replace('and',',',regex=True)
auth['authors'] =auth['authors'].replace('AP','',regex=True)
auth['tokens'] = auth['authors'].str.split(', ')

# Create new columns based on the split values
auth = auth.join(auth['tokens'].apply(lambda x: pd.Series(x)))
auth.drop(columns=['Unnamed: 0'],inplace=True)
auth.to_excel("authorrs2.xlsx")
```
